```python
def remove_negatives(numbers):
    for n in numbers:
        print(f'текущий список: {numbers}, в данный момент {n=}')
        if n < 0:
            numbers.remove(n)
    return numbers

remove_negatives([1, -2, -3, 4, -5])
```

```python
def remove_negatives_fixed(numbers):
    for n in numbers[:]:
        if n < 0:
            numbers.remove(n)
    return numbers

remove_negatives_fixed([1, -2, -3, 4, -5])
```

```python
numbers = [1, -2, -3, 4, -5]
it = iter(numbers)

while True:
    idx = it.__reduce__()[2]
    try:
        n = next(it)
    except StopIteration:
        break
    print(f'индекс {idx=}, взяли {n=}, список сейчас: {numbers=}')
    if n < 0:
        numbers.remove(n)
```
