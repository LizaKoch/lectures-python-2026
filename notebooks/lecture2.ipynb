## Блок 3. Алиасинг и мутабельный аргумент по умолчанию

```python
def append_item(item: str, bucket: list[int] = None) -> list[int]:
    # if bucket is None:
    #     bucket = []
    bucket = bucket or []
    bucket.append(item)
    return bucket

print(append_item(1, [0]))
print(append_item(2))
print(append_item(3))
```

## Блок 4. Поверхностное и глубокое копирование

```python
import copy
```

```python
original = [[1, 2], [3, 4]]
shallow = original[:]  # или shallow2 = list(original) или shallow3 = copy.copy(original)
print(shallow is original)
print(shallow[0] is original[0])

shallow[0].append(99)
print(f'{original}')
```

```python
original = [[1, 2], [3, 4]]
deep = copy.deepcopy(original)

deep[0].append(99)
print(f'{original=}')
print(f'{deep=}')
```

## Блок 5. list vs tuple под капотом

```python
tpl = (1, 2, 3)
print(type(tpl))
```

```python
import sys

lst = [1, 2, 3, 4, 5]
tpl = (1, 2, 3, 4, 5)

print(f'sizeof list: {sys.getsizeof(lst)}')
print(f'sizeof tuple: {sys.getsizeof(tpl)}')
```

```python
l = []

prev = {sys.getsizeof(l)}
print(f'len=0, bytes={prev}')

for i in range(20):
    l.append(i)
    cur = {sys.getsizeof(l)}
    if cur != prev:
        print(f'len={i + 1}, bytes={cur} <<<< память перевыделилась')
        prev = cur
```

```python
import dis

def make_list():
    return [1, 2, 3, 4, 5]

def make_tuple():
    return (1, 2, 3, 4, 5)

print('------LIST------')
print(dis.dis(make_list))
print('------TUPLE------')
print(dis.dis(make_tuple))
```

```python
import timeit

t_l = timeit.timeit("[1, 2, 3, 4, 5]", number=1_000_000)
t_t = timeit.timeit("(1, 2, 3, 4, 5)", number=1_000_000)

print(f'список x1_000_000: {t_l:.4f} сек')
print(f'кортеж x1_000_000: {t_t:.4f} сек')

```

```python
d = {}  # или dict()
d
d[(1, 2)] = 'точка'
print(d)
print(type(d))
```

```python
d = {}  # или dict()
try:
    d[[1, 2]] = 'точка'
except TypeError as e:
    print("Список нельзя использовать в качестве ключа:", e)
print(d)
print(type(d))
```

```python
t = (1, 2, [3, 4])
t[2].append(5)
print(f'после изменения вложенного списка: {t}')

try:
    t[2] = [9, 9]
except TypeError as e:
    print(f'заменить элемент кортежа нельзя', e)
```

```python
from collections import namedtuple

Point = namedtuple("Point", ["x", "y"])
p = Point(1, 2)
print(p)
print(p.x, p.y)

try:
    p.x = 10
except AttributeError as e:
    print('namedtuple неизменяем:', e)
```
