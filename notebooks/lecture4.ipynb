### Группировку

```python
words = ['apple', 'avocado', 'banana', 'bluberry', 'cherry']
res = {}

for w in words:
    if w[0] not in res:
        res[w[0]] = []
    res[w[0]].append(w)
print(res)
```

```python
res = {}
for w in words:
    res.setdefault(w[0], []).append(w)
print(res)
```

```python
from collections import defaultdict, Counter
```

```python
res = defaultdict(list)

for w in words:
    res[w[0]].append(w)
print(res)

print(res['d'])
print('d' in res)
```

```python
class Key:
    def __init__(self, name, h):
        self.name = name
        self.h = h

    def __hash__(self):
        print(f'   hash({self.name}) -> {self.h}')
        return self.h

    def __eq__(self, other):
        print(f'   __eq__: {self.name} == {other.name}?')
        return self.name == other.name

    def __repr__(self):
        return self.name
```

```python
d = {}
print('put А')
d[Key('A', 1)] = 1
print('put B')
d[Key('B', 2)] = 2
print('put C')
d[Key('C', 1)] = 3


print('search C')
print(d[Key('C', 1)])
```

```python
import random, string

def rnd():
    return ''.join(random.choices(string.ascii_lowercase, k=6))

for size in (8, 64, 1024):
    seen = {}
    while True:
        w = rnd()
        i = hash(w) % size
        if i in seen and seen[i] != w:
            print(f'size={size}: {seen[i]!r} и {w!r} -> ячейка {i}, после {len(seen) + 1} строк')
            break
        seen[i] = w
```
